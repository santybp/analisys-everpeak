### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


<details>
<summary>Haz clic para ver la pista</summary>
para eliminar símbolos, puedes reemplazarlos por un texto vacío.

```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic['update_time_utc'] = pd.to_datetime(traffic['update_time_utc'], errors='coerce', utc=True)
traffic["update_time_utc_week_ago"] = pd.to_datetime( traffic["update_time_utc_week_ago"], errors="coerce", utc=True)


# verificar el cambio
traffic.info()
```

```python
# Limpia separadores y convierte columnas numéricas en eco
eco['city_gdp_capita'] = (
    eco['city_gdp_capita'].astype(str)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

eco['unemployment_pct'] = (
    eco['unemployment_pct'].astype(str)
    .str.replace('%', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

eco['population_m'] = (
    eco['population_m'].astype(str)
    .str.replace(',', '.', regex=False)
    .astype(float)
)
# Calcula la población total en unidades absolutas (Multiplica * 1000000)
eco['population'] = eco['population_m'] * 1000000

# verificar el cambio
eco.info()
eco.head(3)
```

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).

- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.

5. RECOMENDACIONES

Bogotá como prioridad de inversión: Los resultados indican que Bogotá debería ser considerada como una ciudad prioritaria para inversión en infraestructura de transporte. Aunque la correlación general entre congestión y productividad es débil, el análisis comparativo de ciudades latinoamericanas revela que Bogotá experimenta niveles de congestión significativamente superiores (41% más que Buenos Aires) con productividad inferior. Esto sugiere que mejoras en movilidad podrían liberar constrains económicos y potencialmente incrementar productividad. La elevada tasa de desempleo (10.0%) respalda esta recomendación, indicando que problemas estructurales de movilidad pueden estar limitando oportunidades laborales.

Validación y profundización: Antes de comprometer recursos en gran escala, se recomienda: (1) Ampliar los datos: Incluir años anteriores (2022-2023) para validar si el patrón es consistente o específico de 2024. (2) Análisis cualitativo: Investigar factores contextuales de Bogotá como estructura urbana, políticas de tránsito, disponibilidad de transporte público y patrones de crecimiento. (3) Modelos de intervención: Evaluar estudios de caso de ciudades que mejoraron movilidad urbana (por ejemplo, Medellín, Curitiba) para identificar lecciones transferibles.

Oportunidades en ciudades emergentes: Ciudades como Lima muestran congestión moderada-alta con PIB por debajo del promedio global, sugiriendo que inversiones estratégicas también podrían resultar efectivas. Sin embargo, el nivel crítico de congestión en Bogotá, combinado con indicadores laborales débiles, justifica priorizarla como caso de estudio inicial.

Limitaciones del análisis: (1) El dataset incluye solo 15 ciudades y un único año, limitando la generalización. (2) La correlación débil indica que la congestión es uno de varios factores que influyen en productividad; otros factores (educación, infraestructura, capital humano) no fueron considerados. (3) Relación entre variables puede ser bidireccional: mayor productividad atrae población y genera tráfico, no necesariamente lo inverso. (4) Los datos de 2024 representan una fotografía temporal; cambios en infraestructura o políticas posteriores no se reflejan.

Próximos pasos: Se recomienda integrar variables adicionales (disponibilidad de transporte público, costo del transporte, tiempo promedio de desplazamiento casa-trabajo), realizar análisis de series temporales si está disponible, y considerar factores geográficos y demográficos que expliquen heterogeneidad entre ciudades. El análisis actual proporciona una base sólida para priorizar Bogotá, pero debe complementarse con investigación cualitativa antes de tomar decisiones de inversión definitivas.

CONCLUSIÓN

La relación entre movilidad urbana y productividad económica es compleja y no es linealmente negativa. Sin embargo, entre las ciudades latinoamericanas analizadas, Bogotá destaca como la ciudad con la mayor necesidad de intervención, combinando altos niveles de congestión con la productividad más baja y desempleo elevado de su grupo de comparación. Aunque los datos no permiten afirmar causalidad, la evidencia sugiere que Bogotá podría ser una ciudad prioritaria para inversión estratégica en infraestructura de transporte, con el potencial de generar retornos económicos significativos. Un análisis más profundo, con datos multi-año y variables contextuales adicionales, es esencial para validar esta recomendación antes de comprometer recursos.
